# 02 — Data Cleaning

Builds on 01_data_loading.ipynb. Filters to London, computes Hi/Ci from TS045, cleans IoD2025
and evse_location (replaces OpenStreetEV_GLA), and cleans the Zapmap session data (dedup +
interval-merge + window-clip so ur_j is mathematically bounded in [0,1]) -- now sourced from
charging_activity_sep.csv (Sept 2025, full month) instead of join_august2025.csv (Aug 4-10).

Two things are structurally new compared to the old pipeline, both because
charging_activity_sep.csv carries no location attributes of its own (unlike join_august2025.csv,
which was pre-joined):
1. Sessions must be explicitly joined to evse_location on `evse_uid` to recover location/LSOA
   context (Section 4).
2. That join must use an EVSE-level (not raw Connector-level) view of evse_location, or every
   session would fan out across however many connectors its EVSE has.

## 1. Reload raw datasets (independent of 01_data_loading.ipynb)

```python
# TS045 (car/van availability by household)
ts045_path = os.path.join(BASE, "03_data/demand/census2021/TS045_car_van_availability_London_LSOA_2021.csv")
ts045_raw = pd.read_csv(ts045_path, skiprows=7).rename(columns={
    "Area": "area_raw",
    "No cars or vans in household": "cars_0",
    "1 car or van in household": "cars_1",
    "2 cars or vans in household": "cars_2",
    "3 or more cars or vans in household": "cars_3plus",
})
ts045_raw = ts045_raw.dropna(subset=["area_raw"])
for col in ["cars_0", "cars_1", "cars_2", "cars_3plus"]:
    ts045_raw[col] = pd.to_numeric(ts045_raw[col], errors="coerce")
ts045_gor_london = ts045_raw[ts045_raw["area_raw"] == "gor:London"].copy()
ts045_lsoa = ts045_raw[ts045_raw["area_raw"].str.startswith("lsoa2021:", na=False)].copy()
parsed = ts045_lsoa["area_raw"].str.extract(r"lsoa2021:(?P<lsoa_code>\S+)\s*:\s*(?P<lsoa_name>.+)")
ts045_lsoa["lsoa_code"] = parsed["lsoa_code"]
ts045_lsoa["lsoa_name"] = parsed["lsoa_name"]
ts045 = ts045_lsoa[["lsoa_code", "lsoa_name", "cars_0", "cars_1", "cars_2", "cars_3plus"]].reset_index(drop=True)

# IoD2025 File 7
iod_raw = pd.read_csv(os.path.join(BASE, "03_data/demand/IoD2025.csv"))
cols_keep = [
    "LSOA code (2021)", "LSOA name (2021)",
    "Local Authority District code (2024)", "Local Authority District name (2024)",
    "Index of Multiple Deprivation (IMD) Score",
    "Index of Multiple Deprivation (IMD) Rank (where 1 is most deprived)",
    "Index of Multiple Deprivation (IMD) Decile (where 1 is most deprived 10% of LSOAs)",
    "Income Score (rate)", "Income Rank (where 1 is most deprived)",
    "Income Decile (where 1 is most deprived 10% of LSOAs)",
]
iod2025 = iod_raw[cols_keep].rename(columns={
    "LSOA code (2021)": "lsoa_code", "LSOA name (2021)": "lsoa_name",
    "Local Authority District code (2024)": "lad_code",
    "Local Authority District name (2024)": "lad_name",
    "Index of Multiple Deprivation (IMD) Score": "imd_score",
    "Index of Multiple Deprivation (IMD) Rank (where 1 is most deprived)": "imd_rank",
    "Index of Multiple Deprivation (IMD) Decile (where 1 is most deprived 10% of LSOAs)": "imd_decile",
    "Income Score (rate)": "income_score",
    "Income Rank (where 1 is most deprived)": "income_rank",
    "Income Decile (where 1 is most deprived 10% of LSOAs)": "income_decile",
})

# evse_location.csv (replaces OpenStreetEV_GLA) -- raw Connector grain, 51,386 rows
evse_location_raw = pd.read_csv(os.path.join(BASE, "03_data/restricted/evse_location.csv"), low_memory=False)
cols_keep = [
    "id_x", "address", "city", "state", "location_class",
    "coordinates_latitude_y", "coordinates_longitude",
    "zapmap_device_uid", "power_band", "device_max_power",
    "evse_id", "evse_uid", "Source", "operator_name",
]
evse_location = evse_location_raw[cols_keep].rename(columns={
    "id_x": "location_id", "state": "borough", "location_class": "location_category",
    "coordinates_latitude_y": "latitude", "coordinates_longitude": "longitude",
    "Source": "source",
})

# charging_activity_sep.csv (replaces join_august2025) -- carries no location attributes
charging_activity_raw = pd.read_csv(os.path.join(BASE, "03_data/restricted/charging_activity_sep.csv"))
charging_activity = charging_activity_raw.rename(columns={
    "start_time": "charging_start", "end_time": "charging_end", "duration": "charging_duration",
})

print("All datasets reloaded.")
print(f"ts045: {ts045.shape}, iod2025: {iod2025.shape}, "
      f"evse_location: {evse_location.shape}, charging_activity: {charging_activity.shape}")
```

## 2. TS045: filter to London + calculate Hi and Ci

```python
london_boroughs = [
    "City of London", "Barking and Dagenham", "Barnet", "Bexley", "Brent",
    "Bromley", "Camden", "Croydon", "Ealing", "Enfield", "Greenwich",
    "Hackney", "Hammersmith and Fulham", "Haringey", "Harrow", "Havering",
    "Hillingdon", "Hounslow", "Islington", "Kensington and Chelsea",
    "Kingston upon Thames", "Lambeth", "Lewisham", "Merton", "Newham",
    "Redbridge", "Richmond upon Thames", "Southwark", "Sutton",
    "Tower Hamlets", "Waltham Forest", "Wandsworth", "Westminster"
]

def is_london(name, boroughs):
    if not isinstance(name, str):
        return False
    return any(re.match(rf"^{re.escape(b)}\s+\d", name) for b in boroughs)

ts045_london = ts045[ts045["lsoa_name"].apply(lambda x: is_london(x, london_boroughs))].reset_index(drop=True)
print("TS045 London rows:", len(ts045_london))
```

```python
band_cols = ["cars_0", "cars_1", "cars_2", "cars_3plus"]
ts045_london["households"] = ts045_london[band_cols].sum(axis=1)
ts045_london["avg_car_ownership"] = (
    0 * ts045_london["cars_0"] + 1 * ts045_london["cars_1"]
    + 2 * ts045_london["cars_2"] + 3 * ts045_london["cars_3plus"]
) / ts045_london["households"]

census_london = ts045_london[["lsoa_code", "lsoa_name", "households", "avg_car_ownership"] + band_cols].copy()
census_london = census_london.rename(columns={"households": "Hi", "avg_car_ownership": "Ci"})

print("=== Census London (TS045-derived) ===")
print("Shape:", census_london.shape)
print(census_london.head(3).to_string())

gor_total_households = ts045_gor_london[band_cols].sum(axis=1).values[0]
computed_total = census_london["Hi"].sum()
print()
print(f"Computed total households: {computed_total:,.0f}")
print(f"Official gor:London aggregate: {gor_total_households:,.0f}")
print(f"Difference: {(computed_total - gor_total_households) / gor_total_households:.4%}")

output_path = os.path.join(BASE, "05_processed/census_london_clean.csv")
census_london.to_csv(output_path, index=False)
print("\nSaved to:", output_path)
```

## 4. Zapmap sessions: datetime parsing, dedup, duration cleaning, join to evse_location

`charging_activity_sep.csv` has no location columns, so before it can be used it needs an
EVSE-level (not raw Connector-level) lookup table to join against -- otherwise every session
would fan out across however many connectors its EVSE happens to have.

```python
# EVSE-level view of evse_location, used only as a join key table for sessions.
# NOT the same as the location-level `osev` built in Section 6 -- this one is deliberately
# kept at evse_uid grain (38,358 rows) so the merge below is 1:1 per session, not fan-out.
evse_location_by_evse = evse_location.drop_duplicates(subset=["evse_uid"])[
    ["evse_uid", "location_id", "latitude", "longitude", "borough", "location_category",
     "power_band", "device_max_power"]
].copy()
print(f"evse_location rows (Connector grain): {len(evse_location):,}")
print(f"evse_location_by_evse rows (EVSE grain): {len(evse_location_by_evse):,}")
```

```python
charging_activity["charging_start"] = pd.to_datetime(charging_activity["charging_start"], errors="coerce")
charging_activity["charging_end"] = pd.to_datetime(charging_activity["charging_end"], errors="coerce")
parse_failures = charging_activity["charging_start"].isna().sum()
print(f"charging_start parse failures: {parse_failures}")

# Drop duplicate session records. Unlike join_august2025.csv (which had a large number of
# exact content-duplicate rows from repeated status-change events, ~70.7% of rows), this file
# is already close to session-grain -- expect a much smaller dedup ratio.
rows_raw = len(charging_activity)
zapmap_dedup = charging_activity.drop_duplicates(subset=["evse_uid", "charging_start", "charging_duration"]).copy()
print(f"Rows before dedup: {rows_raw:,}  |  after: {len(zapmap_dedup):,}  "
      f"|  removed: {rows_raw - len(zapmap_dedup):,} ({(1 - len(zapmap_dedup)/rows_raw):.1%})")

# Duration sanity filter -- window size changed from 10,080 min (7 days) to 43,200 min (30 days)
rows_before = len(zapmap_dedup)
zapmap_dedup = zapmap_dedup[
    (zapmap_dedup["charging_duration"] > 0) &
    (zapmap_dedup["charging_duration"] <= 43200) &
    (zapmap_dedup["charging_start"].notna())
].copy()
print(f"Rows before duration filter: {rows_before:,}  |  after: {len(zapmap_dedup):,}")
```

```python
# Join to recover location/LSOA context -- this step didn't exist in the old pipeline
# because join_august2025.csv already carried location attributes on every row.
zapmap_clean = zapmap_dedup.merge(evse_location_by_evse, on="evse_uid", how="left")

unmatched = zapmap_clean["location_id"].isna().sum()
print(f"Sessions with no matching evse_location row (should be 0): {unmatched:,}")

zapmap_clean["hour"] = zapmap_clean["charging_start"].dt.hour
zapmap_clean["dayofweek"] = zapmap_clean["charging_start"].dt.dayofweek

print("\n=== Zapmap Cleaned ===")
print("Shape:", zapmap_clean.shape)
print("Unique evse_uid:", zapmap_clean["evse_uid"].nunique())
print("Unique location_id:", zapmap_clean["location_id"].nunique())

# Kept as zapmap_clean.csv (same filename as the old pipeline) so 03/04/05 -- not yet
# rewritten -- keep working unchanged until we get to them.
output_path = os.path.join(BASE, "05_processed/zapmap_clean.csv")
zapmap_clean.to_csv(output_path, index=False)
print("\nSaved to:", output_path)
```

## 5. Per-EVSE ur_j: interval-merged, window-clipped

```python
# Merge overlapping session intervals per EVSE (a single long plug-in event can be
# re-logged as several overlapping rows -- confirmed 427 EVSEs / 775 session-pairs have this
# in the new data), clipped to the full September window. Clipping guarantees ur_j <= 1.0
# by construction. Window changed from Aug 4-11 (7 days) to Sept 1 - Oct 1 (30 days).
WINDOW_START = pd.Timestamp("2025-09-01 00:00:00")
WINDOW_END   = WINDOW_START + pd.Timedelta(days=30)
WINDOW_MINUTES = (WINDOW_END - WINDOW_START).total_seconds() / 60

def merged_occupied_minutes(group):
    group = group.sort_values("charging_start")
    total = 0.0
    cur_start = cur_end = None
    for start, dur in zip(group["charging_start"], group["charging_duration"]):
        end = start + pd.Timedelta(minutes=dur)
        if cur_start is None:
            cur_start, cur_end = start, end
        elif start <= cur_end:
            cur_end = max(cur_end, end)
        else:
            cs, ce = max(cur_start, WINDOW_START), min(cur_end, WINDOW_END)
            if ce > cs:
                total += (ce - cs).total_seconds() / 60
            cur_start, cur_end = start, end
    if cur_start is not None:
        cs, ce = max(cur_start, WINDOW_START), min(cur_end, WINDOW_END)
        if ce > cs:
            total += (ce - cs).total_seconds() / 60
    return total

merged_duration = (
    zapmap_clean.groupby("evse_uid")
    .apply(merged_occupied_minutes, include_groups=False)
    .reset_index(name="total_duration_min")
)
evse_coords = zapmap_clean.groupby("evse_uid")[["location_id", "latitude", "longitude"]].first().reset_index()
evse_ur = merged_duration.merge(evse_coords, on="evse_uid", how="left")
evse_ur["ur_j"] = evse_ur["total_duration_min"] / WINDOW_MINUTES

print("=== Per-EVSE ur_j ===")
print("Unique EVSEs with >=1 session:", len(evse_ur))
print(evse_ur["ur_j"].describe())
print(f"EVSEs with ur_j > 1: {(evse_ur['ur_j'] > 1).sum()} (should be 0)")

output_path = os.path.join(BASE, "05_processed/evse_ur_clean.csv")
evse_ur.to_csv(output_path, index=False)
print("\nSaved to:", output_path)
```

## 6. evse_location: de-duplicate to Location level + bounding box check

```python
# evse_location is at Connector grain by construction (one row per connector, so the same
# physical Location repeats many times) -- this is expected, not a data defect, unlike the
# old OpenStreetEV_GLA export where 30 duplicate location_id rows genuinely were a minor
# data-quality issue. Reduce to one row per location_id to reproduce the registry-level table.
print(f"Rows before Location-level reduction: {len(evse_location):,}")
osev_dedup = evse_location.drop_duplicates(subset=["location_id"]).copy()
print(f"Unique locations: {len(osev_dedup):,}")

outside_bbox = osev_dedup[
    (osev_dedup["latitude"] < 51.28) | (osev_dedup["latitude"] > 51.70) |
    (osev_dedup["longitude"] < -0.51) | (osev_dedup["longitude"] > 0.33)
]
print(f"Rows outside London bounding box: {len(outside_bbox)}")

osev_clean = osev_dedup[
    (osev_dedup["latitude"] >= 51.28) & (osev_dedup["latitude"] <= 51.70) &
    (osev_dedup["longitude"] >= -0.51) & (osev_dedup["longitude"] <= 0.33)
].reset_index(drop=True)

print()
print("=== evse_location Cleaned (osev_clean) ===")
print("Shape:", osev_clean.shape)
print("location_category value counts (raw casing):")
print(osev_clean["location_category"].value_counts())
n_onstreet = (osev_clean["location_category"].str.lower() == "on-street").sum()
print(f"On-street locations: {n_onstreet:,}")

output_path = os.path.join(BASE, "05_processed/osev_london_clean.csv")
osev_clean.to_csv(output_path, index=False)
print("\nSaved to:", output_path)
```
