## Load TS045

```python
# Load Census 2021 TS045 (car/van availability by household). Nomis export has 7 metadata
# rows before the header, and a disclosure-control footer with non-numeric values in the
# first count column — both handled explicitly below (TS007A's header was clean; this one
# is not). Replaces TS007A entirely: driving-age population (Pi) is no longer used anywhere
# in the demand formula, which now runs on household vehicle stock (Hi x Ci) instead.
ts045_path = os.path.join(BASE, "03_data/demand/census2021/TS045_car_van_availability_London_LSOA_2021.csv")
ts045_raw = pd.read_csv(ts045_path, skiprows=7)

ts045_raw = ts045_raw.rename(columns={
    "Area": "area_raw",
    "No cars or vans in household": "cars_0",
    "1 car or van in household": "cars_1",
    "2 cars or vans in household": "cars_2",
    "3 or more cars or vans in household": "cars_3plus",
})

# Drop the disclosure-control footer (rows with no area_raw), and coerce count columns to
# numeric -- the footer text in cars_0 forces that column to string dtype otherwise
ts045_raw = ts045_raw.dropna(subset=["area_raw"])
for col in ["cars_0", "cars_1", "cars_2", "cars_3plus"]:
    ts045_raw[col] = pd.to_numeric(ts045_raw[col], errors="coerce")

# Keep the gor:London aggregate row separately -- used in 02_data_cleaning as an independent
# check that the London LSOA subset sums to the correct household total
ts045_gor_london = ts045_raw[ts045_raw["area_raw"] == "gor:London"].copy()

# Parse "lsoa2021:E01000001 : City of London 001A" into lsoa_code + lsoa_name.
# Despite the filename, this file is all England+Wales LSOAs, not pre-filtered to London --
# same as TS007A, the London filter happens in 02_data_cleaning, not here.
ts045_lsoa = ts045_raw[ts045_raw["area_raw"].str.startswith("lsoa2021:", na=False)].copy()
parsed = ts045_lsoa["area_raw"].str.extract(r"lsoa2021:(?P<lsoa_code>\S+)\s*:\s*(?P<lsoa_name>.+)")
ts045_lsoa["lsoa_code"] = parsed["lsoa_code"]
ts045_lsoa["lsoa_name"] = parsed["lsoa_name"]

ts045 = ts045_lsoa[["lsoa_code", "lsoa_name", "cars_0", "cars_1", "cars_2", "cars_3plus"]].reset_index(drop=True)

print("=== TS045 Car/Van Availability (raw, all England+Wales) ===")
print("Shape:", ts045.shape)
print("Columns:", ts045.columns.tolist())
ts045.head()
```

## Load evse_location (replaces OpenStreetEV_GLA)

`OpenStreetEV_GLA.csv` is retired — supervisor-provided `evse_location.csv` covers the full
Location > Device > EVSE > Connector hierarchy in one already-joined file (hence the `_x`/`_y`
suffixed columns), so it fully replaces the old registry's role. Confirmed via a standalone
exploration pass before writing this cell:

- `id_x` (Location id) matches `location_id_x` in every row, and is the correct key for
  LSOA-level eⱼ: 23,013 unique Locations across 51,386 total rows (Location < Device < EVSE <
  Connector counts increase monotonically as expected).
- The usable coordinate pair is `coordinates_latitude_y` + `coordinates_longitude`. The
  Location-level `coordinates_latitude_x` / `coordinates_latitude_x.1` columns are a duplicate-
  latitude artefact in the source file (identical values in both, no true longitude pairing) —
  not used.
- `location_class` values are `On-street` / `Destination` / `En-route` / `Other` — note the
  lowercase "street", unlike the old `On-Street`. Any downstream filter must compare
  case-insensitively.
- `Source` / `operator_name` are kept this time (the old pipeline dropped them as "not used in
  current methodology") because they now feed directly into the L10 data-completeness
  limitation: zero-session EVSEs skew towards `Source == "Custom"` and towards missing
  `operator_name`.

```python
# Load evse_location.csv -- restricted GLA/Zapmap EVSE registry (Sept 2025 snapshot)
# low_memory=False avoids a DtypeWarning on columns we aren't keeping anyway
# (external_evse_uid, status_schedule, directions, parking_restrictions, deleted_at_x)
evse_location_raw = pd.read_csv(os.path.join(BASE, "03_data/restricted/evse_location.csv"), low_memory=False)

# Keep only the columns used in the current methodology; drop near-100%-empty admin/media
# fields (terms_and_conditions, deleted_at_x, floor_level, status_schedule, parking_restrictions,
# directions, images, tariff_ids, suboperator_name, owner_name, format, standard, power_type,
# max_voltage, max_amperage, connector-level ids, timestamps)
cols_keep = [
    "id_x", "address", "city", "state", "location_class",
    "coordinates_latitude_y", "coordinates_longitude",
    "zapmap_device_uid", "power_band", "device_max_power",
    "evse_id", "evse_uid", "Source", "operator_name",
]
evse_location = evse_location_raw[cols_keep].copy()

# Rename to match the old OpenStreetEV_GLA naming convention wherever a direct equivalent
# exists, so downstream 02-05 code needs the smallest possible diff
evse_location = evse_location.rename(columns={
    "id_x":                    "location_id",
    "state":                   "borough",
    "location_class":          "location_category",
    "coordinates_latitude_y":  "latitude",
    "coordinates_longitude":   "longitude",
    "Source":                  "source",
})

print("=== evse_location.csv ===")
print("Shape:", evse_location.shape)
print("Columns:", evse_location.columns.tolist())
evse_location.head()
```

```python
# Integrity checks carried over from the exploration pass -- confirm the two id fields
# that look redundant actually are, before relying on either one downstream
id_check = (evse_location_raw["id_x"] == evse_location_raw["location_id_x"]).all()
evse_check = (evse_location_raw["evse_id"] == evse_location_raw["evse_uid"]).all()
print(f"id_x == location_id_x for all rows: {id_check}")
print(f"evse_id == evse_uid for all rows: {evse_check}")

print("\n=== location_category value counts (raw casing -- compare case-insensitively downstream) ===")
print(evse_location["location_category"].value_counts(dropna=False))
```

## De-duplicate `evse_location` to Location level -> `osev`

```python
# evse_location is at Connector grain (51,386 rows across 23,013 Locations) -- unlike the
# old OpenStreetEV_GLA export, which already was one row per Location. Reduce to one row per
# location_id, keeping the first occurrence's location-level attributes, to reproduce the old
# `osev` registry table used for has_evse_i / eJ.
osev = evse_location.drop_duplicates(subset=["location_id"]).reset_index(drop=True)

n_onstreet = (osev["location_category"].str.lower() == "on-street").sum()
print("=== osev (deduplicated to Location level) ===")
print("Shape:", osev.shape)
print(f"On-street locations: {n_onstreet:,}")
print("(cross-check: the corrected on-street eJ total under the old pipeline was 21,366 --")
print(f" new count is {n_onstreet:,}, consistent within rounding)")
osev.head()
```
