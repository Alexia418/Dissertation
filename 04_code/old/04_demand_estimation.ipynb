# 04 — Demand Estimation

Builds Bᵢ (base demand), then the equity-adjusted Dᵢ(α) with total-demand normalisation,
for the four α scenarios A/B/C/D (α = 0, 0.1, 0.3, 0.5). Consumes 03_EDA's on-street-filtered
eⱼ/ūᵢ/has_evse_i -- no changes needed here versus the earlier (unfiltered) run other than
the input numbers themselves.

## 0. Setup and reload cleaned data

```python
import pandas as pd
import numpy as np
import geopandas as gpd
import matplotlib.pyplot as plt
import os

BASE = "/Users/alexia/Documents/CASA/Dissertation"

census_london = pd.read_csv(os.path.join(BASE, "05_processed/census_london_clean.csv"))
imd_london    = pd.read_csv(os.path.join(BASE, "05_processed/imd_london_clean.csv"))
seff_london   = pd.read_csv(os.path.join(BASE, "05_processed/seff_london.csv"))

figures_dir = os.path.join(BASE, "06_outputs/figures")
os.makedirs(figures_dir, exist_ok=True)

print("Datasets reloaded.")
print(f"census_london: {census_london.shape}, imd_london: {imd_london.shape}, seff_london: {seff_london.shape}")
```

## 1. Base demand Bᵢ = (Hᵢ × Cᵢ) × Pd × Ppub × Uᵢ

```python
Pd = 4.3 / 14
Ppub = 1.8 / 4.4
print(f"Pd   = 4.3 / 14  = {Pd:.6f}")
print(f"Ppub = 1.8 / 4.4 = {Ppub:.6f}")
```

```python
demand_base = census_london[["lsoa_code", "lsoa_name", "Hi", "Ci"]].merge(
    seff_london[["lsoa_code", "ej", "seff", "ubar_i", "has_evse_i"]], on="lsoa_code", how="left"
)
demand_base["Ui"] = np.where(demand_base["has_evse_i"], 1 + demand_base["ubar_i"], 1.0)

print("Ui summary:")
print(demand_base["Ui"].describe())
print(f"\nLSOAs with Ui = 1 exactly: {(demand_base['Ui'] == 1.0).sum()}")
assert demand_base["Ui"].isna().sum() == 0
```

```python
demand_base["Vi"] = demand_base["Hi"] * demand_base["Ci"]
demand_base["Bi"] = demand_base["Vi"] * Pd * Ppub * demand_base["Ui"]

print("=== Bᵢ summary ===")
print(demand_base["Bi"].describe())
```

## 2. Equity-adjusted demand Dᵢ(α), normalised so total demand is constant across scenarios

```python
# === 替换掉 04_demand_estimation.ipynb 里 "## 2. Equity-adjusted demand" 那个cell ===
demand_full = demand_base.merge(imd_london[["lsoa_code", "income_score"]], on="lsoa_code", how="left")
assert demand_full["income_score"].isna().sum() == 0

SCENARIOS = {"A": 0.0, "B": 0.1, "C": 0.3, "D": 0.5}

def demand_for_alpha(Bi, IMDi, alpha):
    # No normalisation: equity weighting is meant to genuinely raise recognised demand in
    # deprived LSOAs, not redistribute a fixed total. Sigma Di therefore grows with alpha.
    return Bi * (1 + alpha * IMDi)

Bi = demand_full["Bi"].to_numpy()
IMDi = demand_full["income_score"].to_numpy()

for label, alpha in SCENARIOS.items():
    demand_full[f"D_{label}"] = demand_for_alpha(Bi, IMDi, alpha)

bi_total = demand_full["Bi"].sum()
print("=== Total demand by scenario (expected to GROW with alpha, not stay constant) ===")
print(f"Sigma Bi: {bi_total:,.2f}")
for label, alpha in SCENARIOS.items():
    total = demand_full[f"D_{label}"].sum()
    pct = (total / bi_total - 1) * 100
    print(f"Sigma D_{label} (alpha={alpha}): {total:,.2f}  ({pct:+.1f}% vs Bi)")
print()
print("alpha=0 (Scenario A) reproduces Bi exactly:", np.allclose(demand_full["D_A"], demand_full["Bi"]))
```

## 3. Distribution comparison and save

```python
scenario_cols = [f"D_{label}" for label in ["A","B","C","D"]]
print(demand_full[scenario_cols].describe())

output_path = os.path.join(BASE, "05_processed/demand_london.csv")
demand_full.to_csv(output_path, index=False)
print(f"\nSaved to: {output_path}")
```

## 4. Sanity check: does higher alpha shift demand toward more deprived LSOAs?

```python
from scipy.stats import pearsonr
print("=== Pearson r: D_scenario vs income_score ===")
for label, alpha in SCENARIOS.items():
    r, p = pearsonr(demand_full[f"D_{label}"], demand_full["income_score"])
    print(f"Scenario {label} (alpha={alpha}): r = {r:.4f}, p = {p:.4g}")
```

## 5. Demand maps — Scenario A (baseline) vs Scenario C (primary)

```python
lsoa_boundaries = gpd.read_file(os.path.join(BASE, "03_data/demand/spatial/LSOA_2021_EW_BGC_V5.shp"))
if lsoa_boundaries.crs is None or lsoa_boundaries.crs.to_epsg() != 27700:
    lsoa_boundaries = lsoa_boundaries.to_crs(epsg=27700)
london_codes = set(census_london["lsoa_code"])
lsoa_london = lsoa_boundaries[lsoa_boundaries["LSOA21CD"].isin(london_codes)].copy()
lsoa_london = lsoa_london.rename(columns={"LSOA21CD": "lsoa_code"})[["lsoa_code", "geometry"]]
print(f"LSOA boundaries matched: {len(lsoa_london)} / {len(london_codes)} expected")
```

```python
# === 替换回共用色阶版本，放在 "## 5. Demand maps" 那个cell ===
demand_map = lsoa_london.merge(demand_full[["lsoa_code", "D_A", "D_B", "D_C", "D_D"]], on="lsoa_code", how="left")
shared_vmax = demand_full[["D_A", "D_B", "D_C", "D_D"]].max().max()
labels = {"A": "alpha=0 (Efficiency)", "B": "alpha=0.1", "C": "alpha=0.3 (Primary)", "D": "alpha=0.5 (Strong Equity)"}

fig, axes = plt.subplots(2, 2, figsize=(16, 16))
for ax, s in zip(axes.flat, ["A", "B", "C", "D"]):
    demand_map.plot(column=f"D_{s}", cmap="viridis", vmin=0, vmax=shared_vmax, linewidth=0.1,
                     edgecolor="grey", legend=True, ax=ax, legend_kwds={"label": "Dᵢ", "shrink": 0.6})
    ax.set_title(f"Scenario {s} — {labels[s]}", fontsize=12)
    ax.axis("off")
plt.suptitle("Equity-Adjusted Demand Across All Four Scenarios (shared colour scale)", fontsize=15, y=0.995)
plt.tight_layout()
plt.savefig(os.path.join(figures_dir, "fig_demand_scenarios_ABCD_shared.png"), dpi=300, bbox_inches="tight")
plt.show()

# === 补充：直接证明"只有高IMD的LSOA才涨得多"这件事，比地图更直观 ===
demand_full["pct_increase_D_vs_A"] = (demand_full["D_D"] / demand_full["D_A"] - 1) * 100

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(demand_full["income_score"], demand_full["pct_increase_D_vs_A"], s=6, alpha=0.3, color="#4C72B0")
ax.set_xlabel("Income Score (IMDᵢ)")
ax.set_ylabel("% increase in demand, Scenario D vs Scenario A")
ax.set_title("Demand Increase Is Concentrated in High-IMD LSOAs, by Design")
plt.tight_layout()
plt.savefig(os.path.join(figures_dir, "fig_demand_increase_vs_imd.png"), dpi=300, bbox_inches="tight")
plt.show()

print(demand_full["pct_increase_D_vs_A"].describe())
```

```python
# === 替换掉 04_demand_estimation.ipynb 里 "## 5. Demand maps" 那个cell ===
demand_map = lsoa_london.merge(demand_full[["lsoa_code", "D_A", "D_B", "D_C", "D_D"]], on="lsoa_code", how="left")
labels = {"A": "alpha=0 (Efficiency)", "B": "alpha=0.1", "C": "alpha=0.3 (Primary)", "D": "alpha=0.5 (Strong Equity)"}

fig, axes = plt.subplots(2, 2, figsize=(16, 16))
for ax, s in zip(axes.flat, ["A", "B", "C", "D"]):
    col = f"D_{s}"
    # 不传 vmin/vmax -- 每张图用自己的min-max，颜色对应的是这张图自己的Di数值
    demand_map.plot(column=col, cmap="viridis", linewidth=0.1, edgecolor="grey",
                     legend=True, ax=ax, legend_kwds={"label": f"Dᵢ (Scenario {s})", "shrink": 0.6})
    ax.set_title(f"Scenario {s} — {labels[s]}\n"
                 f"range: {demand_map[col].min():.1f} – {demand_map[col].max():.1f}", fontsize=12)
    ax.axis("off")
plt.suptitle("Equity-Adjusted Demand by Scenario (each map own colour scale)", fontsize=15, y=0.995)
plt.tight_layout()
plt.savefig(os.path.join(figures_dir, "fig_demand_scenarios_ABCD_ownscale.png"), dpi=300, bbox_inches="tight")
plt.show()
```

```python
# === 04_demand_estimation.ipynb 里，Section 5，替换掉原来 "A vs C" 那个2宫格cell ===
demand_map = lsoa_london.merge(demand_full[["lsoa_code", "D_A", "D_B", "D_C", "D_D"]], on="lsoa_code", how="left")
shared_vmax = demand_full[["D_A", "D_B", "D_C", "D_D"]].max().max()
labels = {"A": "alpha=0 (Efficiency)", "B": "alpha=0.1", "C": "alpha=0.3 (Primary)", "D": "alpha=0.5 (Strong Equity)"}

fig, axes = plt.subplots(2, 2, figsize=(16, 16))
for ax, s in zip(axes.flat, ["A", "B", "C", "D"]):
    demand_map.plot(column=f"D_{s}", cmap="viridis", vmin=0, vmax=shared_vmax, linewidth=0.1,
                     edgecolor="grey", legend=True, ax=ax, legend_kwds={"label": "Dᵢ", "shrink": 0.6})
    ax.set_title(f"Scenario {s} — {labels[s]}", fontsize=12)
    ax.axis("off")
plt.suptitle("Equity-Adjusted Demand Across All Four Scenarios", fontsize=15, y=0.995)
plt.tight_layout()
plt.savefig(os.path.join(figures_dir, "fig_demand_scenarios_ABCD.png"), dpi=300, bbox_inches="tight")
plt.show()
```

```python
# === 04_demand_estimation.ipynb 里，接在原来那张ABCD地图cell之后，作为补充图 ===
demand_map = lsoa_london.merge(demand_full[["lsoa_code", "D_A", "D_B", "D_C", "D_D"]], on="lsoa_code", how="left")
for s in ["B", "C", "D"]:
    demand_map[f"delta_{s}"] = demand_map[f"D_{s}"] - demand_map["D_A"]

shared_vmax_delta = demand_map[[f"delta_{s}" for s in ["B", "C", "D"]]].max().max()

fig, axes = plt.subplots(1, 3, figsize=(20, 8))
labels = {"B": "alpha=0.1", "C": "alpha=0.3 (Primary)", "D": "alpha=0.5 (Strong Equity)"}
for ax, s in zip(axes, ["B", "C", "D"]):
    demand_map.plot(column=f"delta_{s}", cmap="OrRd", vmin=0, vmax=shared_vmax_delta, linewidth=0.1,
                     edgecolor="grey", legend=True, ax=ax, legend_kwds={"label": "Extra demand vs Scenario A", "shrink": 0.6})
    ax.set_title(f"Scenario {s} minus Scenario A ({labels[s]})", fontsize=12)
    ax.axis("off")
plt.suptitle("Where Equity Weighting Adds Recognised Demand (relative to alpha=0 baseline)", fontsize=14, y=1.0)
plt.tight_layout()
plt.savefig(os.path.join(figures_dir, "fig_demand_delta_vs_A.png"), dpi=300, bbox_inches="tight")
plt.show()

print("Delta summary (extra demand vs Scenario A):")
print(demand_map[[f"delta_{s}" for s in ["B", "C", "D"]]].describe())
```
